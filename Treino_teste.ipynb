Exercicio 3.2: Metodo da Bisseção


```python
def f(x):
    return 5*np.exp(-x) + x -5

xs = np.linspace(0,10,1000)

fs = f(xs)

plt.plot(xs,fs)
plt.plot(xs, np.zeros(len(xs)))

def binarySearch(x1,x2,erro):
    
    while(abs(x1-x2)>erro):
        x3 = (x1+x2)/2
     
        if f(x1)*f(x3) <= 0:
            x2 = x3
        else:
            x1 = x3
        

    return (x1 + x2)/2

print(binarySearch(4,6,erro))
```

Exercicio 3.3: Metodo de Newton

```python
def P(x):
    return 924*x**6 - 2772*x**5 + 3150*x**4 - 1680*x**3 + 420*x**2 - 42*x +1

xs = np.linspace(0,1,100)
Ps = P(xs)

plt.plot(xs,Ps)
plt.plot(xs,np.zeros(len(xs)))
plt.show()

def dP(x):
    return 924*6*x**5 - 2772*5*x**4 + 3150*4*x**3 - 1680*3*x**2 + 420*2*x - 42


def newtons(f,df,xinicial,erro):
    x1 = xinicial
    x2 = x1 - (f(x1)/df(x1))
    while(abs(f(x1)-f(x2))> erro):
        x1 = x2
        x2 = x1 - (f(x1)/df(x1))
    
    return x2

print(newtons(P,dP,0.1,1e-10))
print(newtons(P,dP,0.18,1e-10))
print(newtons(P,dP,0.35,1e-10))
print(newtons(P,dP,0.55,1e-10))
print(newtons(P,dP,0.81,1e-10))
print(newtons(P,dP,1,1e-10))
```

```python
def f(x):
    return x**4 - 2*x + 1

a = 0
b = 2
N = 10

def trapezio(f,a,b,N):
    h = abs(b-a)/N
    xs = np.linspace(a,b,N+1)

    return h*(f(xs[0])/2 + f(xs[-1])/2 + np.sum(f(xs[1:-1])))

print(trapezio(f,a,b,N))

def Simpson(f,a,b,N):
    h = abs(b-a)/N
    xs = np.linspace(a,b,N+1)

    return (h/3) *(f(xs[0]) + f(xs[-1]) + 4*np.sum(f(xs[1:-1:2])) + 2*np.sum(f(xs[2:-1:2])))

print(Simpson(f,a,b,N))
```

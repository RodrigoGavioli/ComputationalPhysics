```python
from numpy import zeros,loadtxt
from pylab import plot,xlim,show
from cmath import exp,pi

def dft(y):
    N = len(y)
    c = zeros(N//2+1,complex)
    for k in range(N//2+1):
        for n in range(N):
            c[k] += y[n]*exp(-2j*pi*k*n/N)
    return c
xs = np.linspace(0,2*np.pi,1000)
y = np.cos(xs)
c = dft(y)
plot(abs(c))
xlim(0,500)
show()
```

### Primeiro exercicio 

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def gauss(A,v):
    N = len(A)
    At = np.copy(A)
    vt = np.copy(v)
    pivot = abs(A[0,0])
    lpivot = 0
    for i in range(N):
        test = abs(At[i,0])
        if test>pivot:
            pivot = test
            A[lpivot,:] = At[i,:]
            A[i,:] = At[lpivot,:]
            v[lpivot] = vt[i]
            v[i] = vt[lpivot]
            lpivot = i
    
    for j in range(N):
        divj = A[j,j]
        A[j,:] /= divj
        v[j] /= divj
        for k in range(i, N):
            divk = A[k,i]
            A[k,:] -= divk*A[j,:]
            v[k] -= divk*v[j]





```
